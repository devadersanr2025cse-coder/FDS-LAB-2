```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, classification_report

# Read the dataset from CSV file
df = pd.read_csv("Social_Network_Ads.csv")

print(df.head())
print(df.info())
print(df.describe())

# Use Age and EstimatedSalary as input features
features = df[["Age", "EstimatedSalary"]].values
label = df["Purchased"].values

x_train, x_test, y_train, y_test = train_test_split(
    features, label, test_size=0.2, random_state=42
)

model = LogisticRegression()
model.fit(x_train, y_train)

print("Training Score:", model.score(x_train, y_train))
print("Testing Score:", model.score(x_test, y_test))

predictions = model.predict(x_test)

print("Confusion Matrix:\n", confusion_matrix(y_test, predictions))
print("Classification Report:\n", classification_report(y_test, predictions))

# Predict purchase for a customer aged 30 with estimated salary 87,000
new_customer = np.array([[30, 87000]])
print("Predicted Purchase:", model.predict(new_customer)[0])
```
